```python
import numpy as np

x = np.arange(1, 11)

print ("original", x)
print("media", np.mean(x))
print("mediana", np.median(x))
print("desviación", np.std(x, ddof=1))
print("varianza",np.var(x, ddof=1))
print("rango", np.ptp(x))
```

```python
x = np.arange(1, 11)

print ("datos originales", x)

print ("original +2", x+2)
print ("original *2", x*2)

a = x+2

print ("media", np.mean(a))
print("mediana", np.median(a))
print("desviación", np.std(a, ddof=1))
print("varianza",np.var(a, ddof=1))
print("rango", np.ptp(a))

```
